#### 1. Read in parquet dataset

```python
import pandas as pandas

file_path = "../data/CIC-BCCC-NRC-TabularIoTAttacks-2024.parquet"
dataframe = pandas.read_parquet(file_path)

print(dataframe.shape)
print(dataframe.head())
```

```python
dataframe.columns
```

#### 2. Get Attack Types from unique Attack Name values

```python
dataframe['Attack Name'].unique()
```

#### 3. Check if each Attack Name type has enough values

```python
dataframe['Attack Name'].value_counts()
```

#### 4. Create a small dataset for building model by grouping by Attack Name

```python
small_df = dataframe.groupby('Attack Name').sample(n=100, replace=True, random_state=42)
small_df = small_df.reset_index(drop=True)

print(small_df.shape)
print(small_df.head())
```
