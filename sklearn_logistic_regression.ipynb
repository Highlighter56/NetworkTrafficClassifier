### Check access to all CSVs in Database

```python
from pathlib import Path
import pandas as pd

dataset_path = Path("Dataset")
csv_files = sorted(dataset_path.glob("*.csv"))

print("Number of CSV files:", len(csv_files))

for file in csv_files:
    print(file.name)
```

### Load all CSV files from Dataset

```python
dataframes = []

for file in csv_files:
    temp_df = pd.read_csv(file)
    temp_df["Source File"] = file.stem
    dataframes.append(temp_df)

df_all = pd.concat(dataframes, ignore_index=True)

print("Combined shape:", df_all.shape)
```

```python
df_all["Attack Name"].value_counts()
```

```python
# 1 = Some type of Attack
# 0 = Benign Traffic
# Displays the count of attack data and benign data
df_all["Label"].value_counts()
```

```python
# Prints out the count of null values in the combined Dataset
missing = df_all.isnull().sum()
print(missing[missing > 0].sort_values(ascending=False))
```

```python
# Check for positive and negative infinity values
numeric_all = df_all.select_dtypes(include="number")

infinite_counts = np.isinf(numeric_all).sum()
print(infinite_counts[infinite_counts > 0].sort_values(ascending=False))
```

```python
# Prints out the duplicate count
print("Exact duplicate rows:", df_all.duplicated().sum())
```

```python
print(
    "Duplicates across original columns:",
    df_all.drop(columns="Source File").duplicated().sum()
)
```

```python
original_columns = df_all.columns.drop("Source File")

cross_file_duplicates = (
    df_all.groupby(list(original_columns), dropna=False)["Source File"]
    .nunique()
)

print(
    "Duplicate row groups appearing in multiple files:",
    (cross_file_duplicates > 1).sum()
)
```
